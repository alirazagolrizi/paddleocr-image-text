# Image Text Extraction with PaddleOCR | استخراج متن تصویر با PaddleOCR

Run PaddleOCR on a local image and save recognized text to a text file.

Open the repository README for prerequisites, input assets, limitations, and run instructions. | برای پیش‌نیازها، ورودی‌ها و راهنمای اجرا README مخزن را ببینید.

```python
from paddleocr import PaddleOCR

image_path = "test2.jpg"
output_path = "ocr_result.txt"

ocr = PaddleOCR(lang="en")
result=ocr.predict(image_path)

all_text = []
for res in result:
    for text in res["rec_texts"]:
        text = text.strip()
        if text:
            all_text.append(text)
            
with open(output_path,"w", encoding="utf-8") as file:
    file.write("\n".join(all_text))

print("باموفقیت انجام شد")
print(f"تعداد خطوط پیدا شده {len(all_text)}")
print(f"فایل خروجی{output_path}")
```
